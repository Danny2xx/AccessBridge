## Synthetic headline metric check

Use this small calculation to keep the result definition unambiguous before wiring in real outputs.

```python
def headline_delta(
    baseline_most_deprived_residents: int,
    optimised_most_deprived_residents: int,
) -> dict[str, int]:
    return {
        "baseline_most_deprived_residents": baseline_most_deprived_residents,
        "optimised_most_deprived_residents": optimised_most_deprived_residents,
        "delta_most_deprived_residents": optimised_most_deprived_residents - baseline_most_deprived_residents,
    }


headline_delta(1_250, 2_900)
```
